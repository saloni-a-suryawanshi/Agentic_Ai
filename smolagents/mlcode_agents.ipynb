```python
ml_code_agent = CodeAgent(
    tools=[DuckDuckGoSearchTool()],
    additional_authorized_imports=['pandas','numpy','sklearn','json'],
    model=model
)
```

```python
def generate_ml_code(prompt):
    full_prompt = """
    create a step by step plan to write python code for svm model using sklearn to classify the iris """
    try:
        response = ml_code_agent.run(full_prompt)
        return response
    except Exception as e:
        return f"Error:{str(e)}"
```

```python
iface = gr.Interface(
    fn=generate_ml_code,
    inputs = "text",
    outputs = 'text',
    title = "ML code generator (gemini 2.0 flash)",
    description = "Enter a machine learning coding task"
)
```
