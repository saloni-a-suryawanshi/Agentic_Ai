```python
from crewai import Agent, Task, Crew, Process

#gemini model
MODEL_NAME = "gemini-2.5-flash"

#----------Agents--------

email_author = Agent(
    role = "Email writer",
    goal = "Write 2 cold emails promoting a video-editing tool.",
    backstory = "Expert cold email writer.",
    llm = MODEL_NAME,
    verbose = True,
    use_litellm = True    # important : forces crewai touse Gemini via Litellm
)
marketing_strategist = Agent(
    role="Marketing Strategist",
    goal="Refine and improve email content for engagement and clarity.",
    backstory="Senior marketing strategist.",
    llm=MODEL_NAME,
    verbose=True,
    use_litellm=True
)

content_specialist = Agent(
    role="Content Specialist",
    goal="Polish grammar, clarity, and final formatting.",
    backstory="Professional copy editor.",
    llm=MODEL_NAME,
    verbose=True,
    use_litellm=True
)

# ---------------- Task ----------------
email_task = Task(
    description="""
    1. Write two variations of a cold email promoting a video editing product.
    2. Refine them for clarity, engagement, and professionalism.
    3. Produce polished final versions.
    """,
    expected_output="Two polished final cold email versions.",
    agent=marketing_strategist
)

# ---------------- Crew ----------------
email_crew = Crew(
    agents=[email_author, marketing_strategist, content_specialist],
    tasks=[email_task],
    process=Process.sequential,
    verbose=True
)

# ---------------- Run ----------------
print("\n🚀 Running Crew with Gemini...\n")
final_output = email_crew.kickoff()
print("\n====== FINAL RESULT ======\n")
print(final_output)
```
