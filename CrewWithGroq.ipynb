```python
from crewai import Agent, Task, Crew, Process
from litellm import completion

# 🔥 Tell CrewAI to use LiteLLM always
LITELLM_MODE = True

# Groq model
MODEL_NAME = "groq/llama-3.1-8b-instant"

# ---------------- Agents ----------------

email_author = Agent(
    role="Email Writer",
    goal="Write 2 cold emails promoting a video-editing tool.",
    backstory="Expert cold-email writer.",
    llm=MODEL_NAME,
    verbose=True,
    use_litellm=True           # ← IMPORTANT
)

marketing_strategist = Agent(
    role="Marketing Strategist",
    goal="Refine & improve email content.",
    backstory="Senior marketing expert.",
    llm=MODEL_NAME,
    verbose=True,
    use_litellm=True           # ← IMPORTANT
)

content_specialist = Agent(
    role="Content Specialist",
    goal="Polish grammar, clarity and engagement.",
    backstory="Professional copy editor.",
    llm=MODEL_NAME,
    verbose=True,
    use_litellm=True           # ← IMPORTANT
)

# ---------------- Task ----------------
email_task = Task(
    description="""
    1. Write two variations of a cold email promoting a video editing product.
    2. Refine them for clarity, engagement & professionalism.
    3. Produce polished final versions.
    """,
    expected_output="Two final cold email versions.",
    agent=marketing_strategist
)

# ---------------- Crew ----------------
email_crew = Crew(
    agents=[email_author, marketing_strategist, content_specialist],
    tasks=[email_task],
    process=Process.sequential,
    verbose=True
)

# ---------------- Run ----------------
print("\n🚀 Running Crew...\n")
final_output = email_crew.kickoff()
print("\n====== FINAL RESULT ======\n")
print(final_output)
```
